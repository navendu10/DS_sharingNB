# GenAI Chatbot with Hugging Face Transformers

Two small console chatbots built on pretrained sequence-to-sequence models:

1. **BlenderBot 400M (distilled)** - `facebook/blenderbot-400M-distill`
2. **FLAN-T5 base** - `google/flan-t5-base`

**How to use:** run the cells in order, then chat in the input box that appears under the chat cell. Type `quit`, `exit` or `bye` to stop.

**Requirements:** `pip install transformers torch` (preinstalled on Google Colab). Models are downloaded from the Hugging Face Hub on first use; no token is needed for these public models.


```python
# AutoTokenizer / AutoModelForSeq2SeqLM pick the right classes for any Hugging Face checkpoint name.
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
```

```python
# Model 1: BlenderBot 400M (distilled), a conversational sequence-to-sequence model from Meta.
# The weights are downloaded from the Hugging Face Hub the first time this cell runs.
model_name = "facebook/blenderbot-400M-distill"
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)
tokenizer = AutoTokenizer.from_pretrained(model_name)
```

```python
def chat_with_bot():
  """Simple console chat loop: type a message, the model replies. Type quit, exit or bye to stop."""
  while True:
    input_text = input("You: ")

    if input_text.lower() in ["quit","exit","bye"]:
      print("Chatbot: Goodbye!")
      break

    inputs = tokenizer.encode(input_text, return_tensors = "pt")
    outputs = model.generate(inputs, max_new_tokens = 150)
    response = tokenizer.decode(outputs[0], skip_special_tokens=True).strip()

    print("Chatbot:", response)

chat_with_bot()
```

```python
# Model 2: FLAN-T5 base (Google), an instruction-tuned model; reuses the same tokenizer/model variable names.
model_name = "google/flan-t5-base"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)
```

```python
def chat_with_bot1():
  """Same chat loop as chat_with_bot(), now using whichever model/tokenizer was loaded last (FLAN-T5)."""
  while True:
    input_text = input("You: ")

    if input_text.lower() in ["quit","exit","bye"]:
      print("Chatbot: Goodbye!")
      break

    inputs = tokenizer.encode(input_text, return_tensors = "pt")
    outputs = model.generate(inputs, max_new_tokens = 150)
    response = tokenizer.decode(outputs[0], skip_special_tokens=True).strip()

    print("Chatbot:", response)

chat_with_bot1()
```
